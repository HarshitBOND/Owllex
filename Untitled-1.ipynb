```python
import subprocess

workspace = r"c:\Users\harsh\OneDrive\project_startup\lexvert"
cmd = ["cmd", "/d", "/c", "npm run test:notifications"]

result = subprocess.run(cmd, cwd=workspace, capture_output=True, text=True)
combined_output = (result.stdout or "") + (("\n" if result.stdout and result.stderr else "")) + (result.stderr or "")

exit_code = result.returncode
status = "passed" if exit_code == 0 else "failed"
first_40_lines = "\n".join(combined_output.splitlines()[:40])

print(f"STATUS={status}")
print(f"EXIT_CODE={exit_code}")
print("OUTPUT_FIRST_40_LINES_START")
print(first_40_lines)
print("OUTPUT_FIRST_40_LINES_END")
```
